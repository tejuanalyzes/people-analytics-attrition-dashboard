# **KEY FINDINGS**


1. **Manager** : Manager- related issues and career growth were the most frequently reported exit reasons, followed by workload, relocation, compensation, personal reasons and better oppurtunities.

2. **Compensation** : Compensation-related exits occured across multiple departments and salary levels, Some employees with less than 3 years of tenure at the company,salaries above 30,000 and atleast one promotion still cited compensation as exit reason.

3. **Workload** : Workload-related exits occured among both employees with and without overtime. A subset of employees citing workload also worked more than 35 hours per week along with overtime.

4. **Performance and Engagement** : Performance score and Engagement score among active and exited employees largely remains the same, no substantial difference was observed in the dataset.

5. **Overtime** : Attrition was 20.7% among employees working no overtime and 21.3% among employees who worked overtime.

6. **Gender** : Female employees had an attrition rate of 21.5%, compared with 20.8% among male employees and 14.3% among non-binary employees.

7. **Department** : Finance had the highest department-level attrition rate at approximately 22.6%, while customer support had the lowest at approximately 18.99%.

8. **Promotions** : Attrition was highest among employees with 0-2 promotions, ranging from approximately 20.8% to 21.4%. Attrition was lower among employees with 3 or more promotions, although some promotion group contain extreme values such as 15, 20 and 30 , these require data- quality reviews.

8. **Salary and working conditions** : Average monthly salary, work hours per week, training hours remain broadly similar for exited and active employees.
